Python में Mutable और Immutable का मतलब है कि किसी object को बनने के बाद उसके अंदर की value बदली जा सकती है या नहीं।

1. Mutable

जिस object को बनने के बाद modify/change किया जा सकता है, उसे Mutable कहते हैं।

Examples:

list
dict
set
bytearray
my_list = [10, 20, 30]


my_list[0] = 100


print(my_list)
# [100, 20, 30]

यहाँ list का object वही रहता है, लेकिन उसके अंदर की value बदल गई। इसलिए list mutable है।

2. Immutable

जिस object को बनने के बाद modify/change नहीं किया जा सकता, उसे Immutable कहते हैं।

Examples:

int
float
str
tuple
bool
frozenset
x = 10


x = 20

यहाँ ऐसा नहीं है कि 10 वाले object को बदलकर 20 कर दिया गया। Python ने 20 के लिए नया object reference कर दिया।

String में भी:

name = "Deepak"


name[0] = "R"

यह error देगा:

TypeError: 'str' object does not support item assignment

क्योंकि string immutable है।

आसान याद रखने का तरीका
Type	Mutable?
list	✅ Mutable
dict	✅ Mutable
set	✅ Mutable
int	❌ Immutable
float	❌ Immutable
str	❌ Immutable
tuple	❌ Immutable
bool	❌ Immutable

एक लाइन में:

Mutable = object के अंदर की value बदल सकते हैं।
Immutable = object बनने के बाद उसकी value directly नहीं बदल सकते।

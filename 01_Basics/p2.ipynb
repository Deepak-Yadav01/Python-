### COMMENTS IN PYTHON
In Python, comments are used to explain code or make notes. The Python interpreter ignores comments.

1. Single-line comments

    Use the # symbol.

```python
# This is a single-line comment
print("Hello, World!")
```

You can also place a comment after a statement:

```python
print("Hello, World!")  # This prints a greeting
```

2. Multi-line comments

Using triple quotes (''' or """) (often used as multi-line strings):


```python
"""
This is a multi-line string.
It is sometimes used as a comment,
but it is actually a string object.
"""
print("Hello")
```
